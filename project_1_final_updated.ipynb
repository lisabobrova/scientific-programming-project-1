## part 1. (10 pt) 
Implement a function that uses the bisection method to compute an approximation of the
root of an arbitrary, continuous function g(x).
*   The arguments of the ‘bisection function’ should contain the name the python function implementing g (defined to take a single argument x and to return a single value g(x)), the initial search interval [a, b], the stopping tolerance ϵ > 0, and a maximum number of steps.
Ensure that the chosen initial interval is admissible for the bisection method (g(a) and g(b) have opposite sign).
*   Stop the iteration when the absolute value of g at the approximate root is smaller than ϵ or when the maximum number of iterations is reached.
Scientific Programming Project 1 Numerical optimization
*   The function should return the root and the value of g evaluated at that root, as well as the
number of iterations and the number of times the value of g is evaluated.
Note: the total number of function evaluations is a measure for the computational cost of the method; when different methods need about the same number of iterations to converge, their computational cost might differ.
*   Use your implemented ‘bisection function’ to find the minimum of f(x) = exp(x^2), starting with a non-symmetric initial interval of search around 0. Hint: use the bisection method to find the root of the first derivative!

```python
#q1 part 1 - bisection method

import math

#Same sign function to be used later:
def same_sign(c,d):
    return (c < 0 and d < 0) or (c > 0 and d > 0)

#Bisection function:
#parameters:
#f        - the function to compute the root of
#a        - the left endpoint of the initial search interval [a,b]
#b        - the right endpoint of the initial search interval [a,b]
#tol      - the stopping tolerance ϵ > 0
#max_iter - the maximum number of iterations

def bisection(f, a, b, tol, max_iter): 
    #evaluating f at a and at b
    num_eval = 2      

    #ensuring that the chosen initial interval is admissable for the bisection method
    if same_sign(f(a),f(b)):      
        print("Error: Initial interval is not admissible for the bisection method")
        return 0, 0, 0, 0

    else:    

        #initializing while loop
        iter = 0

        #initial value to start the while loop
        fc = tol+1    
        
        while (iter < max_iter) and (abs(fc) > tol): #end conditions
            #while loop begings, increase the number of iterations 
            iter += 1

            #compute the midpoint at each interval
            c = (a + b) / 2

            #compute f at the midpoin and at a
            fc = f(c)

            #computing at either endpoint does not make a difference
            fa = f(a)

            #increase the number of evaluations by 2
            num_eval += 2


            #chosing the next interval by implementing same sign function:
            if same_sign(fa, fc):
                a = c
            else:
                b = c

    return c, fc, iter, num_eval
    
#Returns:
# c        - the computed root
# fc       - the value of f evaluated at the root 
# iter     - the number of iterations 
# num_eval - the number of times the value of f is evaluated


#function to find root of 
def f(x):   #d/dx(exp(x^2)) -> 0 = 2xexp(x^2)
    return 2 * x * math.exp(x ** 2) 

print("initial search interval [a,b]:")
a = float(input("a = "))
b = float(input("b = "))
tol = float(input("Stopping tolerance = "))
max_iter = float(input("Maximum number of iterations = "))

#calling the bisection function and assigning the values returned to new variables
root, f_root, iterations, evaluations = bisection(f, a, b, tol, max_iter)

print("The approximate root: ", root)
print("The value of f(x) evaluated at the root: ", f_root)
print("The number of iterations: ", iterations)
print("The number of times the value of f(x) is evaluated: ", evaluations)
```

```python
#q1 part 3 - brute force 

%reset -f
import math

#Two functions that will be called in the brute force function, i.e. f(x) and f'(x):

def f(x):
    return math.exp(x ** 2)
def df(x):
    return 2 * x * math.exp(x ** 2)

#Brute force function:
#parameters:
#f        - the function containing f(x)
#df       - the function containing f'(x)
#a0,b0    - the endpoints of the initial search interval [a,b]
#tol      - the tolerance for verifying the value of the derivative
#max_iter - the maximum number of iterations

def brute_force(a0, b0, f, df, tol, max_iter): 

    #first iteration will evaluate the midpoint of the initial search interval (n = number of subintervals = 1)
    n = 1

    #initialise the count for iterations and evaluations
    iter = 0
    num_eval = 0

    #evaluating midpoint of first interval
    mid = (a0 + b0) / 2


    while (iter < max_iter) and (abs(df(mid)) > tol):
        #while loop begings, increase the number of iterations and evaluations
        iter += 1
        num_eval += 1 #dcalling f(mid) in the while loop conditions

        #assigning any value to fmid for first iteration, to initialise while loop 
        #(this value has no meaning as first iteration will be based on i==0 condition in if function below)
        fmid = 0 
        
        #for loop that will compute endpoints of subintervals, find the midpoint of each subinterval, evaluate f and df at the 
        #midpoint of each of these subintervals, reassign the value of fmid, and assign values to dfmid and mid 
        for i in range(0,n):
            #compute the new endpoints of new subintervals
            a = a0 + (b0 - a0) *  i      / n
            b = a0 + (b0 - a0) * (i + 1) / n

            #compute the new midpoint from the new subinterval
            c = (a + b) / 2

            #compute f at each new midpoint
            fc = f(c)
            dfc = df(c)

            #increase the number of evaluations
            num_eval += 2

            if (fc < fmid) or (i == 0): 
                #assign the new values to the variables 
                fmid = fc  #after i==0 fmid changes to fc, which will be used in future comparisons
                dfmid = dfc
                mid = c

        #if the absolute value of the derivative at the approximate minimum  does not yet satisfy the given tolerance
        #double the number of subintervals (n)
        n *= 2    

    return(mid, dfmid, iter, num_eval)

#Returns:
#mid      - the root (position of minimum)
#dfmid    - the value of f' at this root
#iter     - the number of iterations  
#num_eval - the number of function evaluations (which should be the sum of the evaluations of the function f and its derivative f′).

print("Search interval [a0,b0]:")
a0 = float(input("a0 = "))
b0 = float(input("b0 = "))
tol = float(input("Tolerance for verifying the value of the derivative = "))
max_iter = int(input("Maximum number of iterations = "))

#calling the brute force function and assigning the values returned to new variables
root, df_root, iterations, evaluations = brute_force(a0, b0, f, df, tol, max_iter)

print("The root (position of minimum): ", root)
print("The value of f' at root: ", df_root)
print("The number of iterations: ", iterations)
print("The number of function evaluations (sum of the evaluations of f and f'): ", evaluations)
```

Although it is true that the brute force approach is guaranteed to give us the correct value for the minimum (as it check all possible values and finds the correct one), it is much more inefficient compared to the bisection method. The brute force method relies on checking every possiblity for the minimum within the initial interval. It keeps dividing the initial interval into subintervals and finds the midpoint of every subinterval (and the f/df at the midpoints), while the bisection method reduces the original interval and finds the midpoint of the reduced interval.
```python
#q2 part 1 - gradient descent

%reset -f
import math

# Gradient descent function:
# parameters:
# f        - the function f(x) to find local minimum
# df       - the derivative of the function f(x)
# x0       - the initial point for the gradient descent
# tol      - the stopping tolerance on the derivative (ϵ)
# L        - the maximum value of f" (i.e L)
# max_iter - the maximum number of iterations

def gradient_descent(f, df, x0, tol, L, max_iter):                                
    #assigning initial value to x
    x = x0

    #intialising the count for iterations and evaluations
    iter = 0
    num_eval = 0

    #call the function for the derivative of f and assign the return value to a new variable
    dfx = df(x)
    num_eval += 1   #evaluating f′

    #since L does not change (fixed a) we site a before the while loop
    a = 1/L   

    while (iter < max_iter) and (abs(dfx) > tol): #end conditions
        #while loop begings, increase the number of iterations 
        iter += 1

        #reassigning value of x
        x = x - a * dfx

        #computing f' at this new x
        dfx = df(x)
        num_eval += 1    #evaluating f′

    return x, f(x), iter, num_eval

#Returns:
# x         - the root
# f(x)      - the f evaluated at that root
# iter      - the number of iterations
# num_eval  - the number of function evaluations (sum of evaluations of f and f′)


# Sanity check

# Function to find local minimum
def f1(x): 
    return x ** 2

# Derivative of the function f(x)
def df1(x):
    return 2 * x 

x0 = float(input("Initial point x0 = "))
tol = float(input("Stopping tolerance on the derivative (ϵ) = "))
L = 2
max_iter = float(input("Maximum number of iterations =  "))

#calling the gradient descent function and assigning the returned values to new variables
root, f_root, iterations, evaluations = gradient_descent(f1, df1, x0, tol, L, max_iter)

print("The root: ", root) 
print("The f evaluated at the root: ", f_root)
print("The number of iterations: ", iterations)
print("The number of function evaluations (sum of evaluations of f and f′): ", evaluations)
print(" ")

# Exponential function

# Function to find local minimum
def f2(x): 
    return math.exp(x ** 2)

# Derivative of the function f(x)
def df2(x):
    return 2 * x * math.exp(x ** 2)  

# Second derivative of the function f(x)
def ddf2(x):
    return ((4 * x ** 2) + 2) * math.exp(x ** 2) 

x0 = float(input("Initial point x0 = "))
tol = float(input("Stopping tolerance on the derivative (ϵ) = "))
print("second derivative at x0:", abs(ddf2(x0)))
L = float(input('L (where f" < L) = '))
max_iter = float(input("Maximum number of iterations =  "))

#calling the gradient descent function and assigning the returned values to new variables
root, f_root, iterations, evaluations = gradient_descent(f2, df2, x0, tol, L, max_iter)

print("The root:", root) 
print("The f evaluated at the root: ", f_root)
print("The number of iterations: ", iterations)
print("The number of function evaluations (sum of evaluations of f and f′): ", evaluations)
```

```python
#q3 part 4 - gradient descent (with fixed a) on f(x) = tanh(x) cos(exp(x^2)) / (ln(x^2 + 1) exp(−x) + 4)^1/3

%reset -f
import math

# Gradient descent function:
# parameters:
# f                - the function f(x) to find local minimum
# derivative       - the derivative of the function f(x)
# x0               - the initial point for the gradient descent
# h                - the step size      
# tol              - the stopping tolerance on the derivative (ϵ)
# L                - the maximum value of f" (i.e L)
# max_iter         - the maximum number of iterations

def gradient_descent(f, derivative, x0, h, tol, L, max_iter):                                
    #assigning initial value to x
    x = x0

    #initialising the count for iterations and evaluations
    iter = 0
    num_eval = 0

    #call the approximate derivative function and assign the return value to a new variable
    dfx = derivative(f, x, h)
    num_eval += 1   #evaluating f′

    #stating initial value for a
    #since L does not change (fixed a) we site a before the while loop
    a = 1/L

    while (iter < max_iter) and (abs(dfx) > tol): #end conditions
        #while loop begings, increase the number of iterations 
        iter += 1

        #reassigning value of x
        x = x - a * dfx

        #computing f' at this new x
        dfx = derivative(f, x, h)
        num_eval += 1    #evaluating f′

    return x, f(x), iter, num_eval

#Returns:
# x         - the root
# f(x)      - the f evaluated at that root
# iter      - the number of iterations
# num_eval  - the number of function evaluations (sum of evaluations of f and f′)


#function f(x) to find local minimum of
def f(x): 
    return (math.tanh(x) * math.cos(math.exp(x ** 2))) / (math.log(x ** 2 + 1) * math.exp(-x) + 4)**1/3

#approximate derivative of function f(x)
def derivative(f, x, h):
    return (f(x+h) - f(x)) / h

#approxmate second derivative of function f(x)
def second_derivative(derivative, x, h):
    return (derivative(f, x + h, h) - derivative(f, x, h)) / h

print("x0 in [0.6,1.3]:")
x0 = float(input("Initial point x0 = "))
x = x0
h = float(input("h ="))
tol = float(input("Stopping tolerance on the derivative (ϵ) = "))

print("maximum f'' on interval [0.6, 1,3] = ~ 6")
L = 100
print("L = 100")
max_iter = float(input("Maximum number of iterations =  "))

#calling the gradient descent function and assigning the returned values to new variables
root, f_root, iterations, evaluations = gradient_descent(f, derivative, x0, h, tol, L, max_iter)

print("The root:", root) 
print("The f evaluated at the root: ", f_root)
print("The number of iterations: ", iterations)
print("The number of function evaluations (sum of evaluations of f and f′): ", evaluations)
```
